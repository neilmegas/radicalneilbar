# RAPPORT replication quickstart
Downloads the public annual shards, reproduces a party-week count, and preserves stable record links. Counts are retained documents, not total party activity.


```python
import json, urllib.request
import pandas as pd
import matplotlib.pyplot as plt
BASE = 'https://neilmegas.github.io/radicalneilbar/'
index = json.load(urllib.request.urlopen(BASE + 'corpus/index.json'))
records = []
for year in index['years']:
    records.extend(json.load(urllib.request.urlopen(BASE + f'corpus/{year}.json')))
df = pd.DataFrame(records)
df.shape
```

```python
# Example: direct/official records per party and ISO week
direct = df[df['di'] == True].copy()
counts = direct.groupby(['w', 'pn']).size().rename('records').reset_index()
counts.sort_values(['w', 'records'], ascending=[True, False]).head(20)
```

```python
# Reproduce a longitudinal figure for one party
party_id = sorted(df['p'].dropna().unique())[0]
series = df[df['p'].eq(party_id)].groupby('w').size()
ax = series.plot(kind='bar', figsize=(10, 4), title=f'{party_id}: retained records')
ax.set_ylabel('Retained records'); plt.tight_layout()
```
